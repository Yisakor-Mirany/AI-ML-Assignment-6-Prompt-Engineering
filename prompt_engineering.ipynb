This notebook compares a baseline prompt with multiple prompt engineering techniques
to extract structured data from unstructured text.

## Constant Input Text

The following unstructured text is used for all prompt experiments.
The input remains constant to isolate the effect of prompt engineering.

````python
input_text = """
Customer John M. placed an order on March 12, 2024 for a Lenovo ThinkPad X1 priced at $1,349.99.
The order number is ORD-88421. Delivery was scheduled for March 18, 2024.
"""

input_text
````

## Baseline Prompt

**Prompt:**
"Extract the data from the text."

### Baseline Output

John ordered a Lenovo ThinkPad X1 for $1349.99 on March 12, 2024.
Order number ORD-88421, delivery March 18.

**Evaluation:**
- Output is not structured
- No JSON formatting
- Not machine-readable
- Missing explicit field labels

**Score:** 3/10

## Technique 1 – Role Prompting

**Prompt:**

"Act as a Senior Data Analyst responsible for cleaning customer transaction logs.
Extract all relevant structured fields from the following text."

### Output

Customer: John M.  
Product: Lenovo ThinkPad X1  
Price: $1349.99  
Order Number: ORD-88421  
Order Date: March 12, 2024  
Delivery Date: March 18, 2024

**Evaluation:**
- Improved understanding of task
- Fields are identified
- Still not valid JSON
- Inconsistent key naming

**Score:** 6/10

## Technique 2 – Output Formatting Constraint

**Prompt:**

"Extract the following fields from the text and return ONLY valid JSON
using this exact schema:

{
  \"customer_name\": \"\",
  \"product\": \"\",
  \"price_usd\": \"\",
  \"order_number\": \"\",
  \"order_date\": \"\",
  \"delivery_date\": \"\"
}"

### Output

```json
{
  "customer_name": "John M.",
  "product": "Lenovo ThinkPad X1",
  "price_usd": "1349.99",
  "order_number": "ORD-88421",
  "order_date": "March 12, 2024",
  "delivery_date": "March 18, 2024"
}

**Evaluation:**
- Output is valid JSON
- Easy to parse programmatically
- Accurate field extraction

**Score:** 9/10

## Technique 3 – Chain-of-Thought Prompting

**Prompt:**

"Before answering, identify each required field step by step.
Then extract the data and return ONLY valid JSON using the schema below."

### Output

```json
{
  "customer_name": "John M.",
  "product": "Lenovo ThinkPad X1",
  "price_usd": "1349.99",
  "order_number": "ORD-88421",
  "order_date": "March 12, 2024",
  "delivery_date": "March 18, 2024"
}

**Evaluation:**
- Highest accuracy
- No missing fields
- Clear reasoning improves reliability

**Score:** 9.5/10

## Final Optimized Prompt

**Prompt:**

"You are a Senior Data Analyst working on enterprise customer transaction logs.

Instructions:
- Think step by step
- Extract all required fields
- Return ONLY valid JSON
- Use the schema exactly as provided

Schema:
{
  \"customer_name\": \"\",
  \"product\": \"\",
  \"price_usd\": \"\",
  \"order_number\": \"\",
  \"order_date\": \"\",
  \"delivery_date\": \"\"
}"

### Final Output

```json
{
  "customer_name": "John M.",
  "product": "Lenovo ThinkPad X1",
  "price_usd": "1349.99",
  "order_number": "ORD-88421",
  "order_date": "March 12, 2024",
  "delivery_date": "March 18, 2024"
}

**Final Evaluation:**
The combined prompt produces consistent, accurate, and production-ready
structured output suitable for enterprise applications.

**Score:** 10/10

## Comparison Summary

| Baseline | None | 3 | Unstructured and unreliable |

## Key Lesson Learned

Prompt engineering has a major impact on LLM performance. Explicit role
definition, strict output constraints, and structured reasoning dramatically
improve accuracy, consistency, and usability of model outputs.
